```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1. Create a sample numeric dataset
data = {
    'Study_Hours': [2, 3, 4, 5, 6, 7, 8, 9],
    'Attendance': [50, 55, 60, 65, 70, 80, 85, 95],
    'Assignment_Marks': [10, 12, 15, 17, 20, 23, 25, 29],
    'Exam_Marks': [35, 40, 45, 50, 60, 70, 80, 90]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# 2. Apply Standardization
scaler = StandardScaler()
standardized_data = scaler.fit_transform(df)

standardized_df = pd.DataFrame(
    standardized_data,
    columns=df.columns
)

print("\nStandardized Dataset:")
print(standardized_df.round(2))

# 3. Verify mean and standard deviation
print("\nMean of each feature:")
print(standardized_df.mean().round(2))

print("\nStandard deviation of each feature:")
print(standardized_df.std(ddof=0).round(2))

# 4. Apply PCA to reduce 4 features to 2 components
pca = PCA(n_components=2)
pca_data = pca.fit_transform(standardized_data)

# 5. Convert PCA results into a DataFrame
pca_df = pd.DataFrame(
    pca_data,
    columns=['PC1', 'PC2']
)

print("\nPCA Transformed Data:")
print(pca_df.round(2))

# 6. Display explained variance
print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_.round(4))

print("\nTotal Variance Retained:")
print(round(pca.explained_variance_ratio_.sum(), 4))

# 7. Visualize PCA output
plt.figure(figsize=(8, 5))

plt.scatter(
    pca_df['PC1'],
    pca_df['PC2'],
    color='blue',
    s=60
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA: Dimensionality Reduction")
plt.grid(True)
plt.tight_layout()
plt.show()
```
