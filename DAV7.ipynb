```python
cluster1 = [2, 3, 4]
cluster2 = [7, 8, 9]

centroid1 = sum(cluster1) / len(cluster1)
centroid2 = sum(cluster2) / len(cluster2)

print("Cluster 1 Centroid:", centroid1)
print("Cluster 2 Centroid:", centroid2)


def calculate_inertia(cluster, centroid):
    total = 0

    print("\nData Point\tDistance\tSquared Distance")

    for point in cluster:
        distance = point - centroid
        squared_distance = distance ** 2

        print(point, "\t\t", distance, "\t\t", squared_distance)

        total += squared_distance

    return total


inertia1 = calculate_inertia(cluster1, centroid1)
inertia2 = calculate_inertia(cluster2, centroid2)


total_inertia = inertia1 + inertia2

print("\nCluster 1 Inertia:", inertia1)
print("Cluster 2 Inertia:", inertia2)
print("Total Inertia:", total_inertia)
```
